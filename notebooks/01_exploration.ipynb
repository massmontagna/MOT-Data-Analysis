Run the same python code (rows = 3, not 5) for every month to check input data.

```python
import pandas as pd
import os

raw_files = os.listdir("../data/raw/MOT testing data results (2024)")
for file in raw_files :
    print(file)
    df = pd.read_csv(("../data/raw/MOT testing data results (2024)/"+file), nrows=3)
    print(df.shape)
    display(df.head())

```

Now, the following code does the same for the failure data.

```python
import pandas as pd
import os

raw_files = os.listdir("../data/raw/MOT Testing data failure item (2024)")
for file in raw_files :
    print(file)
    df = pd.read_csv(("../data/raw/MOT Testing data failure item (2024)/"+file), nrows=3)
    print(df.shape)
    display(df.head())
```
